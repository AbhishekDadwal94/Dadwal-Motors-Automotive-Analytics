```python
# Build vehicle-type probability function

def get_vehicle_type_weights(province, customer_segment, sales_team):

    province_weights = regional_vehicle_weights[province]

    segment_weights = segment_vehicle_weights[customer_segment]

    team_weights = sales_team_vehicle_weights[sales_team]

    combined_weights = {}

    for vehicle_type in vehicle_types:

        combined_weights[vehicle_type] = (
            province_weights[vehicle_type]
            * segment_weights[vehicle_type]
            * team_weights[vehicle_type]
        )

    total_weight = sum(combined_weights.values())

    normalized_weights = {
        vehicle_type: weight / total_weight
        for vehicle_type, weight in combined_weights.items()
    }

    return normalized_weights


# Test the function with one example

test_weights = get_vehicle_type_weights(
    province="Ontario",
    customer_segment="Retail",
    sales_team="Retail"
)

print("Example vehicle-type probabilities:\n")

for vehicle_type, probability in test_weights.items():
    print(f"{vehicle_type}: {probability:.2%}")

print(f"\nTotal: {sum(test_weights.values()):.2%}")
```

```python
# Assign realistic sales channels

channel_rules = {
    "Retail": {
        "Showroom": 0.65,
        "Online": 0.25,
        "Phone": 0.10
    },
    "Premium": {
        "Showroom": 0.75,
        "Online": 0.15,
        "Phone": 0.10
    },
    "Commercial": {
        "Showroom": 0.35,
        "Online": 0.10,
        "Phone": 0.55
    },
    "Fleet": {
        "Showroom": 0.20,
        "Online": 0.05,
        "Phone": 0.20,
        "Corporate": 0.55
    }
}

def select_sales_channel(segment):

    channels = list(channel_rules[segment].keys())
    probabilities = list(channel_rules[segment].values())

    return np.random.choice(
        channels,
        p=probabilities
    )


transaction_base["SalesChannel"] = [
    select_sales_channel(segment)
    for segment in transaction_base["CustomerSegment"]
]

print("Sales channels assigned.")

print("\nOverall channel distribution:")
print(
    transaction_base["SalesChannel"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nChannel distribution by customer segment:")
display(
    pd.crosstab(
        transaction_base["CustomerSegment"],
        transaction_base["SalesChannel"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)
```

```python
# Assign realistic payment methods

payment_rules = {
    "Retail": {
        "Financing": 0.55,
        "Cash": 0.25,
        "Lease": 0.20
    },
    "Premium": {
        "Financing": 0.45,
        "Lease": 0.35,
        "Cash": 0.20
    },
    "Commercial": {
        "Financing": 0.50,
        "Cash": 0.30,
        "Lease": 0.20
    },
    "Fleet": {
        "Financing": 0.45,
        "Lease": 0.45,
        "Cash": 0.10
    }
}

def select_payment_method(segment):

    methods = list(payment_rules[segment].keys())
    probabilities = list(payment_rules[segment].values())

    return np.random.choice(
        methods,
        p=probabilities
    )


transaction_base["PaymentMethod"] = [
    select_payment_method(segment)
    for segment in transaction_base["CustomerSegment"]
]

print("Payment methods assigned.")

print("\nOverall payment method distribution:")
print(
    transaction_base["PaymentMethod"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nPayment method distribution by customer segment:")
display(
    pd.crosstab(
        transaction_base["CustomerSegment"],
        transaction_base["PaymentMethod"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)
```

```python
# Assign payment term

def select_payment_term(payment_method):

    if payment_method == "Cash":
        return 0

    elif payment_method == "Financing":
        return np.random.choice(
            [36, 48, 60, 72],
            p=[0.15, 0.30, 0.40, 0.15]
        )

    elif payment_method == "Lease":
        return np.random.choice(
            [24, 36, 48],
            p=[0.25, 0.55, 0.20]
        )


transaction_base["PaymentTermMonths"] = [
    select_payment_term(payment_method)
    for payment_method in transaction_base["PaymentMethod"]
]

print("Payment terms assigned.")

print("\nOverall payment term distribution:")
print(
    transaction_base["PaymentTermMonths"]
    .value_counts()
    .sort_index()
)

print("\nPayment term by payment method:")
display(
    transaction_base
    .groupby("PaymentMethod")["PaymentTermMonths"]
    .agg(
        Count="count",
        AverageTerm="mean",
        MinimumTerm="min",
        MaximumTerm="max"
    )
    .round(2)
)
```

```python
# Create DimDate

date_range = pd.date_range(
    start="2022-01-01",
    end="2025-12-31",
    freq="D"
)

dim_date = pd.DataFrame({
    "Date": date_range
})

dim_date["Year"] = (
    dim_date["Date"].dt.year
)

dim_date["MonthNumber"] = (
    dim_date["Date"].dt.month
)

dim_date["MonthName"] = (
    dim_date["Date"].dt.month_name()
)

dim_date["MonthShortName"] = (
    dim_date["Date"].dt.strftime("%b")
)

dim_date["QuarterNumber"] = (
    dim_date["Date"].dt.quarter
)

dim_date["Quarter"] = (
    "Q"
    + dim_date["QuarterNumber"].astype(str)
)

dim_date["YearMonth"] = (
    dim_date["Date"].dt.strftime("%Y-%m")
)

dim_date["YearQuarter"] = (
    dim_date["Year"].astype(str)
    + "-Q"
    + dim_date["QuarterNumber"].astype(str)
)

dim_date["MonthStart"] = (
    dim_date["Date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

dim_date["DayOfMonth"] = (
    dim_date["Date"].dt.day
)

dim_date["DayOfWeekNumber"] = (
    dim_date["Date"].dt.dayofweek + 1
)

dim_date["DayOfWeekName"] = (
    dim_date["Date"].dt.day_name()
)

dim_date["IsWeekend"] = (
    dim_date["Date"].dt.dayofweek >= 5
)

print("DimDate created.")

print(f"Rows: {len(dim_date):,}")

print(
    f"Minimum date: "
    f"{dim_date['Date'].min().date()}"
)

print(
    f"Maximum date: "
    f"{dim_date['Date'].max().date()}"
)

print(
    f"Unique dates: "
    f"{dim_date['Date'].nunique():,}"
)

print(
    f"Duplicate dates: "
    f"{dim_date['Date'].duplicated().sum():,}"
)

print(
    f"Missing dates: "
    f"{dim_date['Date'].isna().sum():,}"
)

print("\nDays by year:")

display(
    dim_date
    .groupby("Year")
    .size()
    .rename("Days")
    .to_frame()
)

print("\nSample:")

display(dim_date.head(10))
```

```python
# Create FactSales
#
# Keep transaction-level business metrics and foreign keys.
# Descriptive attributes belong in the dimension tables.

fact_columns = [
    # Transaction identifiers
    "SalesID",
    "SalesDate",

    # Dimension keys
    "CustomerID",
    "DealershipID",
    "SalespersonID",
    "VehicleID",

    # Transaction attributes
    "SalesChannel",
    "PaymentMethod",
    "PaymentTermMonths",

    # Customer lifecycle
    "CustomerPurchaseNumber",
    "CustomerPurchaseType",
    "DaysSincePreviousPurchase",
    "PurchaseIntervalBand",
    "CustomerTenureMonths",
    "CustomerTenureBand",

    # Financial inputs
    "Quantity",
    "MSRP",
    "ActualCost",

    # Pricing / discount
    "BaseDiscountRate",
    "PromotionAdjustment",
    "FinalDiscountRate",
    "MarketAdjustmentRate",
    "SalePrice",

    # Financial outputs
    "Revenue",
    "GrossProfit",
    "GrossMarginRate",

    # Modeling drivers
    "GrowthFactor",
    "SeasonalityFactor",
    "MonthStart"
]

fact_sales = (
    transaction_base[fact_columns]
    .copy()
)

print("=" * 60)
print("FACT SALES CREATED")
print("=" * 60)

print(f"\nRows: {len(fact_sales):,}")

print(
    f"Unique SalesIDs: "
    f"{fact_sales['SalesID'].nunique():,}"
)

print(
    f"Duplicate SalesIDs: "
    f"{fact_sales['SalesID'].duplicated().sum():,}"
)

print(
    f"Missing values: "
    f"{fact_sales.isna().sum().sum():,}"
)

print("\nForeign-key coverage:")

print(
    f"Customers:   "
    f"{fact_sales['CustomerID'].nunique():,}"
)

print(
    f"Dealerships: "
    f"{fact_sales['DealershipID'].nunique():,}"
)

print(
    f"Salespeople: "
    f"{fact_sales['SalespersonID'].nunique():,}"
)

print(
    f"Vehicles:    "
    f"{fact_sales['VehicleID'].nunique():,}"
)

print("\nFinancial totals:")

print(
    f"Revenue:      "
    f"${fact_sales['Revenue'].sum():,.0f}"
)

print(
    f"Gross Profit: "
    f"${fact_sales['GrossProfit'].sum():,.0f}"
)

print(
    f"Average Margin: "
    f"{fact_sales['GrossMarginRate'].mean():.2%}"
)

print("\nSample:")

display(
    fact_sales.head(10)
)
```

```python
# Validate FactSales foreign keys against dimension tables

print("=" * 70)
print("FOREIGN KEY REFERENTIAL INTEGRITY")
print("=" * 70)


def validate_foreign_key(
    fact_df,
    fact_key,
    dimension_df,
    dimension_key,
    dimension_name
):

    fact_values = set(
        fact_df[fact_key].dropna().unique()
    )

    dimension_values = set(
        dimension_df[dimension_key].dropna().unique()
    )

    orphan_values = fact_values - dimension_values

    orphan_rows = (
        fact_df[fact_df[fact_key].isin(orphan_values)]
        .shape[0]
    )

    print(
        f"\n{fact_key} → {dimension_name}"
    )

    print(
        f"Dimension keys: {len(dimension_values):,}"
    )

    print(
        f"Orphan key values: {len(orphan_values):,}"
    )

    print(
        f"Orphan fact rows: {orphan_rows:,}"
    )

    return orphan_rows


orphan_customer = validate_foreign_key(
    fact_sales,
    "CustomerID",
    dim_customer,
    "CustomerID",
    "DimCustomer"
)

orphan_vehicle = validate_foreign_key(
    fact_sales,
    "VehicleID",
    dim_vehicle,
    "VehicleID",
    "DimVehicle"
)

orphan_dealership = validate_foreign_key(
    fact_sales,
    "DealershipID",
    dim_dealership,
    "DealershipID",
    "DimDealership"
)

orphan_salesperson = validate_foreign_key(
    fact_sales,
    "SalespersonID",
    dim_salesperson,
    "SalespersonID",
    "DimSalesperson"
)

orphan_dates = validate_foreign_key(
    fact_sales,
    "SalesDate",
    dim_date,
    "Date",
    "DimDate"
)

print("\n" + "=" * 70)

total_orphans = (
    orphan_customer
    + orphan_vehicle
    + orphan_dealership
    + orphan_salesperson
    + orphan_dates
)

print(
    f"Total orphan fact rows: {total_orphans:,}"
)

print("=" * 70)
```

```python
# ============================================================
# FINAL DATASET FREEZE VALIDATION
# ============================================================

print("=" * 70)
print("FINAL DATASET FREEZE VALIDATION")
print("=" * 70)

expected_counts = {
    "FactSales": 531_482,
    "DimCustomer": 125_000,
    "DimVehicle": 25,
    "DimDealership": 10,
    "DimSalesperson": 80,
    "DimDate": 1_461
}

actual_counts = {
    "FactSales": len(fact_sales),
    "DimCustomer": len(dim_customer),
    "DimVehicle": len(dim_vehicle),
    "DimDealership": len(dim_dealership),
    "DimSalesperson": len(dim_salesperson),
    "DimDate": len(dim_date)
}

print("\nRow-count validation:")

all_counts_correct = True

for table_name, expected in expected_counts.items():

    actual = actual_counts[table_name]

    status = "PASS" if actual == expected else "FAIL"

    print(
        f"{table_name:<18} "
        f"Expected: {expected:>10,} | "
        f"Actual: {actual:>10,} | "
        f"{status}"
    )

    if actual != expected:
        all_counts_correct = False


print("\nKey validation:")

checks = {
    "FactSales unique SalesID":
        fact_sales["SalesID"].nunique() == len(fact_sales),

    "DimCustomer unique CustomerID":
        dim_customer["CustomerID"].nunique() == len(dim_customer),

    "DimVehicle unique VehicleID":
        dim_vehicle["VehicleID"].nunique() == len(dim_vehicle),

    "DimDealership unique DealershipID":
        dim_dealership["DealershipID"].nunique() == len(dim_dealership),

    "DimSalesperson unique SalespersonID":
        dim_salesperson["SalespersonID"].nunique() == len(dim_salesperson),

    "DimDate unique Date":
        dim_date["Date"].nunique() == len(dim_date)
}

all_keys_correct = True

for check_name, result in checks.items():

    print(
        f"{check_name:<40} "
        f"{'PASS' if result else 'FAIL'}"
    )

    if not result:
        all_keys_correct = False


print("\nFinancial validation:")

revenue = fact_sales["Revenue"].sum()
gross_profit = fact_sales["GrossProfit"].sum()
avg_margin = fact_sales["GrossMarginRate"].mean()

print(f"Revenue:      ${revenue:,.0f}")
print(f"Gross Profit: ${gross_profit:,.0f}")
print(f"Average Margin: {avg_margin:.2%}")


print("\n" + "=" * 70)

if all_counts_correct and all_keys_correct:
    print("DATASET STATUS: FROZEN")
    print("No structural validation failures detected.")
else:
    print("DATASET STATUS: NOT FROZEN")
    print("Resolve validation failures before exporting.")

print("=" * 70)
```
